# 動態 window 內按日分批買入
2023-03-01 至 2026-10-14，每月入金1萬元，合計44萬元。維持前一版動態diff與低水準條件；首次進入時於收盤鎖定當時現金C與window天數W，從下一可交易日開始，每天投入C/W，連續W個0050可交易日。條件之後不成立仍繼續既定計畫。沒有10萬元總額上限。

同一計畫不疊加；期間新入金留待下一輪；未在計畫進行時出現新觸發就不開新計畫。訊號若在最後一筆成交當天收盤出現，可開新輪。零現金不開計畫。每日金額含0.1425%買入費率，使用可分割總報酬等值單位，未計最低手續費、零股價差與股息稅；0050分割及配息沿用已核對調整行情。末日若未完成則保留剩餘現金，本次所有計畫都完成。
